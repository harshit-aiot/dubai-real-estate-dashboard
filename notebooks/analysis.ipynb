```python
# Dubai Real Estate Intelligence Dashboard
# Data Analysis Notebook

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("C:\harshitlab\data anlys\dubai housing  projrct\housing_price_dataset.csv")

# Display Dataset
print(df.head())

# Dataset Information
print(df.info())

# Check Missing Values
print(df.isnull().sum())

# Remove Duplicates
df = df.drop_duplicates()

# Statistical Summary
print(df.describe())

# Average Property Price by Location
location_price = df.groupby("Neighborhood")["Price"].mean().sort_values(ascending=False)

print(location_price)

# Top 10 Expensive Locations
top_locations = location_price.head(10)

plt.figure(figsize=(10,6))
top_locations.plot(kind='bar')

plt.title("Top 10 Expensive Locations")
plt.xlabel("Location")
plt.ylabel("Average Price")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Property Type Distribution
property_distribution = df["Neighborhood"].value_counts()


plt.figure(figsize=(7,7))
property_distribution.plot(kind='pie', autopct='%1.1f%%')

plt.title("Property Category Distribution")
plt.ylabel("")

plt.show()

# Square Feet vs Price
plt.figure(figsize=(8,5))

plt.scatter(df["SquareFeet"], df["Price"])

plt.title("Property Size vs Price")
plt.xlabel("Square Feet")
plt.ylabel("Price")

plt.show()

# Average Price Per Neighborhood
avg_price = df.groupby("Neighborhood")["Price"].mean()

plt.figure(figsize=(8,5))
avg_price.plot(kind='bar')

plt.title("Average Price by Neighborhood")
plt.xlabel("Neighborhood")
plt.ylabel("Average Price")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Correlation Matrix
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

print(correlation)

# Final Insights
print("Insights Generated Successfully")
```
